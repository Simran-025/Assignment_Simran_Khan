```python
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "meta-llama/Llama-3.2-1B-Instruct"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

print("Tokenizer:\n", tokenizer)
print("Model:\n", model_name)
```

```python
from transformers import pipeline
generator = pipeline("text-generation", model=model, tokenizer=tokenizer)
```

```python
role=input("Enter role: ")
ques=input("enter your ques: ")

msg=[
    {
    "role":"system",
    "content":(f"You are a healthcare assistant in a {role}."
               "Provide helpful and structured medical advise.")
    },
    {
        "role":"user",
        "content":ques
    }
]
prompt=tokenizer.apply_chat_template(
        msg,
        tokenize=False,
        add_generation_prompt=True
    )
output=generator(prompt, max_new_tokens=200, pad_token_id=tokenizer.eos_token_id)[0]["generated_text"]
assistant=output[len(prompt):]
print("\nHuman:  ", ques)
print("\nAssistant: ",assistant)
```

```python
#Task-2
prompt="""
you are an experienced doctor
A 60 year old lady came want your advice to control her diabetes
Explain using some aspects given below:
1) Disease Overview
2) Symptoms
3) Risk Factors
4) Prevention
5) When to Consult a Doctor

Use simple language so a non-medico also understand.
"""

response=generator(prompt, max_new_tokens=200, pad_token_id=tokenizer.eos_token_id)[0]["generated_text"]
assistant = response[len(prompt):].strip()
print("\nHuman:  ", prompt)
print("\nAssistant: ",assistant)
```

```python
#Few Shot Prompt

prompt="""
you are an experienced doctor

Below are examples of converting patient complaints into structured summaries.

Example 1:
Patients has symptomps of vomitting, stomach ache and weakness

Disease: Possible Diarrhea

Symptoms:
Vomitting
Fever
Weakness
stomach ache

Recommendation:
Drink plenty of water
Eat light food like khichdi and fruits
Avoid oily and spicy food
Consult a doctor for proper medication

Example 2

Clinical Note:
Patient has cough, sore throat, and mild fever.

Structured Summary:

Disease: Possible Viral Infection

Symptoms:
Cough
Sore throat
Mild fever

Recommendation:
Drink warm fluids.
Take adequate rest.
Gargle with warm salt water.
Consult a doctor if symptoms persist.


Now convert below Note into the same structured format.

Note:
A 60-year-old woman has high blood sugar, frequent urination, excessive thirst, and fatigue.



Use simple language so a non-medico also understand.
"""

response=generator(prompt, max_new_tokens=500, pad_token_id=tokenizer.eos_token_id)[0]["generated_text"]
assistant = response[len(prompt):].strip()
print("\nHuman:  ", prompt)
print("\nAssistant: ",assistant)
```

```python
## Chain of thought

prompt="""
You are an experienced doctor.
A 60 year old lady came want your advice to control her diabetes.

Before providing your answer
1) carefully understand the medical condition.
2) Decide most appropriate medical recommendation.
3) Consider the possible causes.

Provide your final answer using the following headings:

1) Disease Overview
2) Possible Cause
3) Symptoms
4) Risk Factors
5) Prevention
6) When to Consult a Doctor

Use simple word so a non-medico also understand.
"""

response=generator(prompt, max_new_tokens=500, pad_token_id=tokenizer.eos_token_id)[0]["generated_text"]
assistant = response[len(prompt):].strip()
print("\nHuman:  ", prompt)
print("\nAssistant: ",assistant)
```

```python
## JSON prompt

prompt="""
You are an experienced doctor.
A patient comes to you with the following symptoms:
High fever, Cough, Body pain and Fatigue

Return a valid JSON object in the following format.

{
  "disease": "",
  "symptoms": [],
  "risk_level": "",
  "recommendation": ""
}

"""

response=generator(prompt, max_new_tokens=150, pad_token_id=tokenizer.eos_token_id)[0]["generated_text"]
assistant = response[len(prompt):].strip()
print("\nHuman:  ", prompt)
print("\nAssistant: ",assistant)
```
